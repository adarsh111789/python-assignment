```python
# 1) CSV File for Address Book
# Import CSV module
import csv

# Open CSV file in write mode
file = open("address_book.csv", "w", newline='')

# Create CSV writer object
writer = csv.writer(file)

# Write column headings
writer.writerow(["Name", "Address", "Mobile", "Email"])

# Taking 3 records from user
for i in range(3):
    print("\nEnter Details of Person", i + 1)

    name = input("Enter Name: ")
    address = input("Enter Address: ")
    mobile = input("Enter Mobile Number: ")
    email = input("Enter Email ID: ")

    # Write data into CSV file
    writer.writerow([name, address, mobile, email])

print("\nData inserted successfully into CSV file.")

# Close file
file.close()
```

```python
#2) DATABASE Practice using SQLite3
#Step 1: Create Database and Tables
import sqlite3

# Connect to database
conn = sqlite3.connect("college.db")

# Create cursor object
cur = conn.cursor()

# Create Student table
cur.execute("""
CREATE TABLE IF NOT EXISTS Student(
    id INTEGER PRIMARY KEY,
    name TEXT,
    branch TEXT
)
""")

# Create Faculty table
cur.execute("""
CREATE TABLE IF NOT EXISTS Faculty(
    fid INTEGER PRIMARY KEY,
    fname TEXT,
    subject TEXT
)
""")

# Create Course table
cur.execute("""
CREATE TABLE IF NOT EXISTS Course(
    cid INTEGER PRIMARY KEY,
    cname TEXT,
    duration TEXT
)
""")

print("Tables created successfully.")

conn.commit()
```
